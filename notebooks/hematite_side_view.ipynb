# Hematite Fe₂O₃(012)-(1×1) side view

Notebook version of the `hematite_side_view.py` example, using the project-specific settings in `profile.py`.

```python
from pathlib import Path

from IPython.display import Image, display

from atomic_povray import (
    Camera,
    Color,
    DepthShading,
    DisplayBounds,
    RenderConfig,
    StyleConfig,
    apply_styles,
    build_geometry,
    get_default_bonds,
    load_structure,
    make_scene,
    render_scene,
    write_ini,
    write_scene,
)
from profile import HEMATITE_SIDE_VIEW, get_side_view_light

project = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
poscar = project / "tests" / "data" / "fe2o3-012-1x1-relaxed.vasp"
povray = r"C:\Program Files\POV-Ray\v3.7\bin\pvengine64.exe"
```

```python
structure = load_structure(poscar)
bond_rules = get_default_bonds(structure, profile=HEMATITE_SIDE_VIEW)
```

```python
geometry = build_geometry(
    structure,
    bounds=DisplayBounds(
        fractional_ranges=((-2.0, 2.0), (-1.5, 1.5), (0.45, 0.75)),
    ),
    bond_rules=bond_rules,
)

print(
    f"{len(geometry.primary_atoms)} primary atoms, "
    f"{len(geometry.extension_atoms)} bond-extension atoms, "
    f"{len(geometry.bonds)} bonds"
)
```

```python
styles = StyleConfig(
    depth_shading=DepthShading(
        origin=(0.0, 0.0, 24.0),
        direction=(0.0, 0.0, -1.0),
        decay_length=30.0,
        target=Color(1.0, 1.0, 1.0),
    ),
    profile=HEMATITE_SIDE_VIEW,
)
styled = apply_styles(geometry, styles)
```

```python
camera = Camera.orthographic(
    direction=(0.0, 100.0, 0.0),
    target=(0.0, 0.0, 21.0),
    profile=HEMATITE_SIDE_VIEW,
)
scene = make_scene(
    styled.primitives,
    camera=camera,
    lights=(get_side_view_light(camera),),
    profile=HEMATITE_SIDE_VIEW,
)
```

```python
render_config = RenderConfig(
    width=1024,
    height=768,
    antialias_threshold=0.05,
    sampling_method=2,
    display_gamma=2.0,
    file_gamma=2.0,
    transparent=True,
    display=True,
    executable=povray,
    profile=HEMATITE_SIDE_VIEW,
)
result = render_scene(
    scene,
    project / "hematite_side_view.png",
    render_config,
)
print(result.image_path)
display(Image(filename=result.image_path))
```

## Export without rendering

Render the `.ini` file in the POV-Ray GUI so that its resolution, gamma, antialiasing, transparency, and quality settings are retained.

```python
scene_path = write_scene(
    scene,
    project / "hematite_side_view.pov",
    width=render_config.width,
    height=render_config.height,
    povray_version=render_config.povray_version,
)
ini_path = write_ini(
    scene_path,
    project / "hematite_side_view.png",
    render_config,
    filename=project / "hematite_side_view.ini",
)
scene_path, ini_path
```
